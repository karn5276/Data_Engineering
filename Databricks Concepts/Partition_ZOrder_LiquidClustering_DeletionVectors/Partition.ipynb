```python
df = spark.sql("select * from samples.wanderbricks.employees")
```

```python
df.write.partitionBy("country").mode("overwrite").saveAsTable("workspace.default.employees_partitioned")
```

```python
%sql describe detail employees_partitioned
```

```python
%sql
select distinct country from employees_partitioned;
```

```python
# below query will run fastly beacuase we are searching rows based on partition key , due to that it read only one numFiles.
%sql
select * from employees_partitioned where country= "United States"
```

```python
%sql
select * from employees_partitioned where role="owner"
```

```python
%sql
select distinct role from employees_partitioned where country="United States";
```
