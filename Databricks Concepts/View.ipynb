
_view is created to perform query fast
eg: if perform join operation on main table again and again then it can increase query time but 
if we create view and store join query result in that then it will not take that much of time._

```python
%sql
select
  *
from
  samples.wanderbricks.employees
```

```python
%sql
select * from samples.wanderbricks.hosts
```

```python
employees_df = spark.table("samples.wanderbricks.employees")
hosts_df = spark.table("samples.wanderbricks.hosts")

joined_df = employees_df.join(hosts_df, employees_df["host_id"] == hosts_df["id"], "inner")

joined_df.createOrReplaceTempView("employees_hosts_view")
```

```python
%sql
CREATE OR REPLACE TEMP VIEW employees_hosts_view AS
SELECT 
    e.name,
    e.country,
    e.employee_id, 
    e.role,
    h.email
FROM samples.wanderbricks.employees e
INNER JOIN samples.wanderbricks.hosts h
ON e.host_id = h.host_id;
```

```python
%sql
select * from employees_hosts_view
```
