```python
%sql
select department,
max(case when rank_no=2 then salary 
when rank_no !<1 then null END) as second_highest_salary
 from (select *, dense_rank() OVER (PARTITION BY department ORDER BY salary DESC) as rank_no from demo.default.employees2)
group by department;
```
