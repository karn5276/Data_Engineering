```python
# 1. clear()

# Removes all items

d = {"a": 1, "b": 2}
d.clear()
print(d)  # {}

# 2. copy()
# Returns shallow copy

d = {"a": 1}
x = d.copy()
print(x)  # {'a': 1}

# 3. fromkeys()
# Creates dictionary from keys

keys = ["a", "b", "c"]
d = dict.fromkeys(keys, 0)
print(d)  # {'a': 0, 'b': 0, 'c': 0}

# 4. get()
# Returns value of key

d = {"a": 1}
print(d.get("a"))      # 1
print(d.get("b", 0))   # 0 , if b exists with different value then it print that value
# dict1[j] = dict1.get(j, 0) + 1  IMPORTANT


# 5. items()
# Returns key-value pairs

d = {"a": 1, "b": 2}
print(d.items())  # dict_items([('a',1),('b',2)])

# 6. keys()
# Returns all keys

d = {"a": 1, "b": 2}
print(d.keys())  # dict_keys(['a','b'])

# 7. values()
# Returns all values

d = {"a": 1, "b": 2}
print(d.values())  # dict_values([1,2])

# 8. pop()
# Removes key and returns value

d = {"a": 1, "b": 2}
d.pop("a")
print(d)  # {'b': 2}

# 9. popitem()
# Removes last inserted item

d = {"a": 1, "b": 2}
d.popitem()
print(d)  # {'a': 1}

# 10. setdefault()
# Returns value, inserts if not exists

d = {"a": 1}
d.setdefault("b", 2)
print(d)  # {'a':1,'b':2}

# 11. update()
# Updates dictionary

d = {"a": 1}
d.update({"b": 2})
print(d)  # {'a':1,'b':2}

# 12. len()
# Returns number of items

d = {"a": 1, "b": 2}
print(len(d))  # 2

# 13. in (membership check)

d = {"a": 1}
print("a" in d)  # True
```
