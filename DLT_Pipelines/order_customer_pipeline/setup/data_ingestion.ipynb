```python
data = [
    (1, "Amit", "amit@gmail.com", "Pune", "2024-01-01"),
    (2, "Rahul", "rahul@gmail.com", "Mumbai", "2024-01-01"),
    (3, "Sneha", "sneha@gmail.com", "Delhi", "2024-01-01"),
]

columns = ["customer_id", "name", "email", "city", "updated_at"]

df = spark.createDataFrame(data, columns)
df.write.format("delta").mode("overwrite").saveAsTable("customer_orders.bronze.new_customers")
```
