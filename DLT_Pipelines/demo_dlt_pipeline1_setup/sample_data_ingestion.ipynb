```python
data = [
    (1, "Rahul", "Pune"),
    (2, "Amit", "Mumbai"),
    (3, "Sneha", "Delhi")
]

columns = ["customer_id", "name", "city"]

df = spark.createDataFrame(data, columns)
df.write.format("delta").mode("overwrite").saveAsTable("demo.bronze.bronze_customers")
```

```python
data = [
    (101, 1, 500, "2024-04-01"),
    (102, 2, 700, "2024-04-02"),
    (103, 1, 300, "2024-04-03")
]

columns = ["order_id", "customer_id", "amount", "order_date"]

df = spark.createDataFrame(data, columns)
df.write.format("delta").mode("overwrite").saveAsTable("demo.bronze.bronze_orders")
```
