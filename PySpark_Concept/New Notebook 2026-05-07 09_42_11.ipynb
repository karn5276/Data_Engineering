```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

data = [
    (1, "Karan", 25, "Pune", 50000),          # normal row
    (2, "Rahul", 30, "Mumbai", 60000),        # normal row
    (3, None, 28, "Delhi", 55000),            # null in name column
    (4, "Amit", None, "Chennai", 45000),      # null in age column
    (5, "Neha", 26, None, 52000),             # null in city column
    (6, "Priya", 29, "Hyderabad", None),      # null in salary column
    (7, None, None, None, None),              # entire row mostly null
    (8, "", 32, "Pune", 58000),               # empty string
    (9, "Suresh", -1, "Bangalore", 62000),    # invalid age
    (10, "Ankit", 27, "Pune", 0),             # invalid salary
    (11, "Karan", 25, "Pune", 50000),         # duplicate row
    (11, "Karan", 25, "Pune", 50000),         # exact duplicate row
    (12, "Rahul", 30, "Mumbai", 60000),       # duplicate business data
    (13, "Rahul", 30, "Mumbai", 60000),       # duplicate business data
    (14, "Priya", 29, "Hyderabad", None),     # partial duplicate
    (15, "Priya", 29, "Hyderabad", None),     # partial duplicate
    (16, "John@", 35, "Delhi", 70000),        # special character in name
    (17, "Meena", 200, "Pune", 40000),        # unrealistic age
    (18, "Ravi", 24, "Pune", -5000),          # negative salary
    (19, "  ", 31, "Nagpur", 48000),          # spaces only
    (20, "Sneha", None, None, None),          # multiple nulls
    (None, None, None, None, None)            # complete null row
]

columns = ["id", "name", "age", "city", "salary"]

df = spark.createDataFrame(data, columns)

df.show(100,truncate=False)
```

```python
df.dropna().show()  # this will drop row if any value in that row in NULL
```

```python
df.dropna(how='all').show(100)  # drop entire row if entire row is null
```

```python
df.dropna(subset=['name','salary']).show()  # it will remove row where null value in name or salary.
```

### Filling Nulls

```python
df.printSchema()
```

```python
df.fillna(5777).show()  # it will fill all only numeric null value to 5777
```

```python
df.fillna('UNKNOWN').show()  # it will fill all only string null value to UNKNOWN
```

```python
df.fillna({'salary':-1,'name':'UNKNOWN','age':18,'city':'UNKNOWN'}).show() # per-columns fill value
```
