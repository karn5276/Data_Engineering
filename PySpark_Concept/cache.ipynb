```python
data = [(1, "A", 100),
        (2, "B", 200),
        (3, "C", 300),
        (4, "D", 400)]

df = spark.createDataFrame(data, ["id", "name", "salary"])
df.show()
```

```python
df_filtered = df.filter(df.salary > 150)

df_filtered.count()   # Action 1
df_filtered.show()    # Action 2
```

```python
df_filtered = df.filter(df.salary > 150)

df_filtered.cache()   # Store in memory

df_filtered.count()   # First time compute + cache
df_filtered.show()    # Second time use cache (FAST)
```

```python
df_filtered = df.filter(df.salary > 150)

df_filtered.createOrReplaceTempView("temp_table")

spark.sql("SELECT COUNT(*) FROM temp_table").show()
spark.sql("SELECT * FROM temp_table").show()
```
