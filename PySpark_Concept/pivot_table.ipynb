Pivot in Spark is used to convert row values into columns.

Pivot always needs:

groupBy(),
pivot(),
aggregation function


```python
data = [
    ("A", "Jan", 100),
    ("A", "Feb", 200),
    ("B", "Jan", 150),
    ("B", "Feb", 300)
]

columns = ["employee", "month", "sales"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
pivot_df = df.groupBy("employee") \
             .pivot("month") \
             .sum("sales")

pivot_df.show()
```
